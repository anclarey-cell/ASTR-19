```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
```

```python
def first_model(x, a, b):
    return  a* np.exp(-b*x)

def second_model(x, a, b):
    return a / (x**b)
```

```python
a = 2
b = 3
x = np.linspace(0, 2*np.pi, 100)
```

```python
y = first_model(x,a,b)
plt.figure(figsize=(8,5))
plt.errorbar(x, y, label = (f"y = {a} e^{{-{b}x}}"), color='blue', fmt = 'o')

popt, pcov = curve_fit(first_model, x, y)
plt.plot(x, first_model(x, *popt), 'r-')

r = second_model(x,a,b)
plt.figure(figsize=(8,5))
plt.errorbar(x, r, label = (f"y = {a} / (x,{b})"), color='green', fmt = '*')

yo = curve_fit(second_model, x, r)
plt.plot(x, second_model(x, *yo), 'r-')


plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.title("X vs. Y", fontsize=14)
plt.legend()
plt.grid(True)
plt.show
```
